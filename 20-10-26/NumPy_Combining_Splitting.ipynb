```python
#Cell 1: Concatenate Arrays
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print(arr1)
print(arr2)

print("Horizontal:", np.concatenate((arr1, arr2)))
print("Vertical:", np.concatenate((arr1, arr2), axis=0))
```

```python
#Cell 2: hstack() અને vstack()
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

print(arr1)
print(arr2)

horizontal_stack = np.hstack((arr1, arr2))
vertical_stack = np.vstack((arr1, arr2))

print("Horizontal Stack:")
print(horizontal_stack)

print("Vertical Stack:")
print(vertical_stack)
```

```python
#Cell 3: vsplit() અને hsplit()
for part in np.vsplit(vertical_stack, 2):
    print(part)

for part in np.hsplit(horizontal_stack, 2):
    print(part)
```

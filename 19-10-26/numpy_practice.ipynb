```python
#Practice 1: Identity Matrix
# Create Matrix 4 x 4
import numpy as np
arr = np.eye(4)
print(arr)
```

```python
#Practice 2: Random Matrix
# create matrix 3 x 3 random integers between 1 to 50
arr = np.random.randint(1, 50, size=(3,3), dtype="int")
print(arr)
```

```python
#Practice 3: Filtering
# create a 1D array of 1 to 20 elements: values greater than 10
arr = np.arange(1, 21)
print(arr)

filtered_arr =arr[arr > 10]
print(filtered_arr)
```

```python
#Practice 4: Even Numbers Filter
# create an array from 1 to 30
arr = np.arange(1,31)
even_arr = arr[arr % 2 == 0]
print(even_arr)
```

```python
#Practice 5: Values Less Than 10
# create an array from 1 to 15
arr = np.arange(1,16)
filtered_arr = arr[arr < 10]
print(filtered_arr)
```

```python
#6: Filter Numbers Greater Than 15
# create an array from 1 to 30
arr = np.arange(1,31)
filtered_arr = arr[arr > 15]
print(filtered_arr)
```

```python
#7:Filter Odd Numbers
# create a array 1 to 20 odd number 
arr = np.arange(1,20)
odd_arr = arr[arr % 2 !=0]
print(odd_arr)
```

```python
#8 Find Numbers Divisible by 3
# create a array 1 to 30 find number divisible by 3 
arr = np.arange(1,31)
divisible_arr = arr [arr % 3 == 0]
print(divisible_arr) 
```

```python
#9 Replace Values
# array values 0 replace
arr = np.array([5, 12, 3, 18, 7, 20, 9])
arr[arr < 10] =0
print(arr)
```

```python
#10 Count Values
# array 10 high values
arr = np.array([4, 15, 8, 22, 11, 3, 19, 6])
count = np.sum(arr > 10)
print("Count:", count)
```

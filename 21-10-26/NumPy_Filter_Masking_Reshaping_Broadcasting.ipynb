```python
# Numpy : Filtering , Masking , Reshaping and Broadcasing

#Program 1: Basic Filtering

import numpy as np
ages  = np.array([18 , 30 , 12 , 35 , 22 , 40 , 17])

adults = ages[ages >= 18]

print(ages)

print(adults)
```

```python
#Program 2: Multiple Conditional Filtering
# Multiple conditional filtering

ages = np.array([18, 30, 12, 35, 22, 40, 17])

young_adult = ages[(ages > 18) & (ages < 35)]

print(young_adult)
```

```python
#Program 3: String Filtering
# String Filtering

names = np.array(["Rahul" , "Arpit" , "Benjamin" , "Alice" , "Zeel" , "Priya" , "Pooja"])

# string start with A and P

selected_values = names[np.char.startswith(names , ["A"])]

print(selected_values)
```

```python
#Program 4: Basic Masking for Selection
# Basic Masking for Selection

temperature = np.array([28, 32, 25, 35, 30, 32])

# True for temperature above 30

is_hot = temperature > 30

hot_temperature = temperature[is_hot]

print("Temperature:", temperature)
print(is_hot)
print(hot_temperature)
```

```python
#Program 5: Using np.where() for Masking
# Using np.where() for masking

scores = np.array([65, 80, 72, 90, 55, 75])

grades = np.where(scores >= 75, "Pass", "Fail")

print("Scores:", scores)
print("Grades:", grades)
```

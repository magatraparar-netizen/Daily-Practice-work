```python
# Search  , Sort  , Aggregation / Statistical Functions

#Program 1: Find Minimum and Maximum Value
# Find Minimum and Maximum value in Numpy Arrays

numbers = np.array([28 , 8 , 42 , 17 , 3 , 31])
print(numbers)
```

```python
print("Maximum :" , np.max(numbers))
print("Minimum:" , np.min(numbers))
print("Index of Maximum:" , np.argmax(numbers))
print("Index of minimum:" , np.argmin(numbers))
```

```python
#Program 2: Sort a 2D Array Along Different Axis
# Sort a 2D array along different axis

matrix = np.array([
    [9, 2, 7],
    [4, 8, 1],
    [6, 3, 5]
])
print(matrix)
print("Sort with axis=0\n", np.sort(matrix, axis=0))

print("Sort with axis=1\n", np.sort(matrix, axis=1))
```

```python
#Program 3: Find Mean, Median and Standard Deviation
# Find mean, median and standard deviation

data = np.random.randint(1, 101, size=9)

print(data)

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Std Dev:", np.std(data))
```

```python
#Program 4: Count Positive, Negative and Zero Values
# Count Positive, Negative and Zero values

values = np.array([-5, 0, 6, 12, -3, 0, -1, 4, 0])

positive_value_count = np.sum(values > 0)
negative_value_count = np.sum(values < 0)
zero_value_count = np.sum(values == 0)

print("Positive_Count:", positive_value_count)
print("Negative_Count:", negative_value_count)
print("Zero_Count:", zero_value_count)
```

```python
#Program 5: Search Elements Using np.where()
# Search elements in NumPy array

numbers = np.array([10, 25, 30, 45, 50, 25])

print("Numbers:", numbers)

print("Index of 25:", np.where(numbers == 25))

print("Values greater than 30:", numbers[numbers > 30])
```
